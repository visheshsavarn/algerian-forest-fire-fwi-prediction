```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
df.head()
```

```python
df.info()
```

# Data Cleaning

```python
df[df.isnull().any(axis=1)]
```

```python
df.loc[:122,'Region']=0
df.loc[122:,'Region']=1
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
#Removing the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

## Fix spaces in column names

```python
df.columns = df.columns.str.strip()
df.columns
```

## Change the required columns as integer data types

```python
df[['day','month','year','Temperature','RH','Ws']] = df[['day','month','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

## Changing the other columns as float data types

```python
cols = [c for c in df.columns
        if pd.api.types.is_string_dtype(df[c]) and c != "Classes"]
df[cols]=df[cols].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.head()
```

### Saving the cleaned dataset

```python
df.to_csv('Algerian_forest_fires_cleaned_dataset_UPDATE.csv')
```
